### Data Preprocessing and Missing Value Handling

This is a critical section where we prepare our data for the VAE model. Key points:

**Missing Value Conventions:**
- In CSV files: `-1` indicates missing values
- In mask files: `True` = missing, `False` = observed
- For PyTorch models: We convert to `1` = observed, `0` = missing (standard convention)

**Why This Matters:**
Our VAE needs to distinguish between observed values (which provide constraints) and missing values (which need to be generated). The mask tells the model which values to trust and which to predict.

```python
# Correcting skeweed distributions
class FeatureTransformer:
    """
    Handles preprocessing for car design features, specifically applying the 
    Yeo-Johnson transformation to skewed features (28-34) to make them more Gaussian.
    
    The transformer is fitted only on non-missing data to ensure robustness.
    """
    def __init__(self, feature_indices_to_transform: List[int]):
        """
        Initializes the transformer.
        :param feature_indices_to_transform: A list of 0-indexed column indices 
                                             corresponding to the skewed features.
        """
        self.indices = feature_indices_to_transform
        # Use a list of transformers, one for each feature
        self.transformers: List[PowerTransformer] = [
            PowerTransformer(method='yeo-johnson', standardize=True) 
            for _ in feature_indices_to_transform
        ]
        print(f"Transformer initialized for {len(self.indices)} features.")

    def fit(self, X_train_NaN: pd.DataFrame):
        """
        Fits the PowerTransformer to the non-missing training data.
        
        :param X_train_NaN: The training data as a DataFrame, with missing values 
                            represented by NaN.
        """
        if not self.indices:
            print("No features configured for transformation.")
            return

        print(f"Fitting PowerTransformer on non-missing data for indices: {self.indices}...")
        
        # 1. Select the columns that need transformation
        cols_to_transform = X_train_NaN.iloc[:, self.indices]
        
        # 2. Iterate through each feature and fit its dedicated transformer
        for i, col_index in enumerate(self.indices):
            feature_data = cols_to_transform.iloc[:, i].dropna().values.reshape(-1, 1)
            
            if len(feature_data) > 0:
                self.transformers[i].fit(feature_data)
                
        print("✓ PowerTransformer fitted successfully.")

    def transform(self, X_NaN: Union[pd.DataFrame, np.ndarray]) -> np.ndarray:
        """
        Applies the fitted transformation to the input data.
        
        :param X_NaN: The data (train, validation, or test) with NaN for missing values.
        :return: A NumPy array with the specified features transformed.
        """
        if not self.indices:
            return X_NaN.values if isinstance(X_NaN, pd.DataFrame) else X_NaN
            
        X_out = X_NaN.values.copy() if isinstance(X_NaN, pd.DataFrame) else X_NaN.copy()
        
        print(f"Transforming data for indices: {self.indices}...")
        
        for i, col_index in enumerate(self.indices):
            # Extract the column data (including NaNs)
            col_data = X_out[:, col_index].reshape(-1, 1)
            
            # Identify where the NaNs are
            nan_mask = np.isnan(col_data).flatten()
            
            # Only transform non-NaN values
            if np.sum(~nan_mask) > 0:
                # Apply the transformation
                transformed_values = self.transformers[i].transform(col_data[~nan_mask])
                
                # Place transformed values back into the output array
                X_out[~nan_mask, col_index] = transformed_values.flatten()

        return X_out
    
```

```python
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset
from sklearn.preprocessing import StandardScaler, QuantileTransformer
from sklearn.metrics import mean_squared_error
import lightgbm as lgb
from tqdm import tqdm
import matplotlib.pyplot as plt
import math  # <--- Added missing import
import warnings

# Import your utils
from utils import load_dataset_splits, plot_distribution_comparison, plot_prediction_scatter

warnings.filterwarnings('ignore')

# ==========================================
# 1. MANIFOLD AUTOENCODER (The Refiner)
# ==========================================
class Autoencoder(nn.Module):
    def __init__(self, input_dim, hidden_dims=[128, 64, 32]):
        super().__init__()
        encoder_layers = []
        in_dim = input_dim
        for h in hidden_dims:
            encoder_layers.append(nn.Linear(in_dim, h))
            encoder_layers.append(nn.BatchNorm1d(h))
            encoder_layers.append(nn.LeakyReLU())
            in_dim = h
        self.encoder = nn.Sequential(*encoder_layers)
        
        decoder_layers = []
        for h in reversed(hidden_dims):
            decoder_layers.append(nn.Linear(in_dim, h))
            decoder_layers.append(nn.BatchNorm1d(h))
            decoder_layers.append(nn.LeakyReLU())
            in_dim = h
        self.decoder = nn.Sequential(*decoder_layers)
        self.final = nn.Linear(in_dim, input_dim)

    def forward(self, x):
        z = self.encoder(x)
        recon = self.decoder(z)
        return self.final(recon)

class Refiner:
    def __init__(self, input_dim, device):
        self.device = device
        self.model = Autoencoder(input_dim).to(device)
        self.scaler = StandardScaler()
        
    def fit(self, data, epochs=100):
        print("Training Autoencoder Refiner...")
        data_norm = self.scaler.fit_transform(data)
        loader = DataLoader(TensorDataset(torch.FloatTensor(data_norm)), batch_size=64, shuffle=True)
        optimizer = optim.Adam(self.model.parameters(), lr=1e-3)
        
        for epoch in range(epochs):
            self.model.train()
            for batch in loader:
                x = batch[0].to(self.device)
                noise = torch.randn_like(x) * 0.05 # Denoising training
                x_noisy = x + noise
                recon = self.model(x_noisy)
                loss = nn.MSELoss()(recon, x)
                optimizer.zero_grad()
                loss.backward()
                optimizer.step()

    def refine(self, samples):
        print("Refining Samples...")
        n_samples, n_variations, n_features = samples.shape
        refined = np.zeros_like(samples)
        self.model.eval()
        
        # Process variations in batches
        flat_samples = samples.reshape(-1, n_features)
        
        batch_size = 1000
        for i in range(0, len(flat_samples), batch_size):
            batch = flat_samples[i:i+batch_size]
            batch_norm = self.scaler.transform(batch)
            batch_tensor = torch.FloatTensor(batch_norm).to(self.device)
            
            with torch.no_grad():
                recon = self.model(batch_tensor)
                # Blend: 70% Original (Signal) + 30% Refined (Structure)
                final = (0.7 * batch_tensor) + (0.3 * recon)
                
            refined_np = self.scaler.inverse_transform(final.cpu().numpy())
            flat_samples[i:i+batch_size] = refined_np
            
        return flat_samples.reshape(n_samples, n_variations, n_features)

# ==========================================
# 2. RESIDUAL DIFFUSION (The Noise Maker)
# ==========================================
class SinusoidalPositionEmbeddings(nn.Module):
    def __init__(self, dim):
        super().__init__()
        self.dim = dim
    def forward(self, time):
        device = time.device
        half_dim = self.dim // 2
        embeddings = math.log(10000) / (half_dim - 1)
        embeddings = torch.exp(torch.arange(half_dim, device=device) * -embeddings)
        embeddings = time[:, None] * embeddings[None, :]
        embeddings = torch.cat((embeddings.sin(), embeddings.cos()), dim=-1)
        return embeddings

class ResidualDiffusion(nn.Module):
    def __init__(self, input_dim, hidden_dims=[512, 512, 512]):
        super().__init__()
        self.input_dim = input_dim
        time_dim = 64
        self.time_mlp = nn.Sequential(
            SinusoidalPositionEmbeddings(time_dim),
            nn.Linear(time_dim, time_dim),
            nn.SiLU(),
            nn.Linear(time_dim, time_dim)
        )
        in_dim = (input_dim * 2) + time_dim
        self.layers = nn.ModuleList()
        for h_dim in hidden_dims:
            self.layers.append(nn.Sequential(
                nn.Linear(in_dim, h_dim),
                nn.BatchNorm1d(h_dim),
                nn.SiLU(),
                nn.Dropout(0.1)
            ))
            in_dim = h_dim
        self.final = nn.Linear(in_dim, input_dim)

    def forward(self, x, t, cond):
        t_emb = self.time_mlp(t)
        x = torch.cat([x, cond, t_emb], dim=1)
        for layer in self.layers: x = layer(x)
        return self.final(x)

class DiffusionManager:
    def __init__(self, model, device, timesteps=300):
        self.model = model
        self.device = device
        self.timesteps = timesteps
        beta_start = 0.0001
        beta_end = 0.02
        self.betas = torch.linspace(beta_start, beta_end, timesteps).to(device)
        self.alphas = 1. - self.betas
        self.alphas_cumprod = torch.cumprod(self.alphas, dim=0)
        self.sqrt_alphas_cumprod = torch.sqrt(self.alphas_cumprod)
        self.sqrt_one_minus_alphas_cumprod = torch.sqrt(1. - self.alphas_cumprod)

    def add_noise(self, x_0, t):
        noise = torch.randn_like(x_0)
        sqrt_alpha = self.sqrt_alphas_cumprod[t][:, None]
        sqrt_one_minus = self.sqrt_one_minus_alphas_cumprod[t][:, None]
        return sqrt_alpha * x_0 + sqrt_one_minus * noise, noise

    def train_step(self, res, cond, optimizer):
        self.model.train()
        optimizer.zero_grad()
        t = torch.randint(0, self.timesteps, (res.shape[0],), device=self.device).long()
        x_t, noise = self.add_noise(res, t)
        pred_noise = self.model(x_t, t, cond)
        loss = nn.MSELoss()(pred_noise, noise)
        loss.backward()
        optimizer.step()
        return loss.item()

    @torch.no_grad()
    def generate(self, cond, n_samples=1):
        self.model.eval()
        n_rows, n_feats = cond.shape
        cond_rep = cond.repeat_interleave(n_samples, dim=0)
        x = torch.randn_like(cond_rep)
        
        for i in reversed(range(self.timesteps)):
            t = torch.full((x.shape[0],), i, device=self.device, dtype=torch.long)
            pred_noise = self.model(x, t, cond_rep)
            beta = self.betas[i]
            alpha = self.alphas[i]
            alpha_bar = self.alphas_cumprod[i]
            mean = (1 / torch.sqrt(alpha)) * (x - (beta / torch.sqrt(1 - alpha_bar)) * pred_noise)
            if i > 0:
                noise = torch.randn_like(x)
                sigma = torch.sqrt(beta)
                x = mean + sigma * noise
            else:
                x = mean
        return x.reshape(n_rows, n_samples, n_feats)

# ==========================================
# 3. TRI-STAGE PIPELINE
# ==========================================
class TriStageImputer:
    def __init__(self, device):
        self.device = device
        self.gbdt_models = {}
        self.diffusion = None
        self.diff_manager = None
        self.refiner = None
        self.res_scaler = QuantileTransformer(output_distribution='normal', n_quantiles=2000)
        self.cond_scaler = StandardScaler()
        self.discrete_vals = {}
        
    def fit(self, data, mask):
        n_samples, n_features = data.shape
        
        # 1. Memorize Discrete
        for i in range(n_features):
            unique = np.unique(data[:, i])
            if len(unique) < 100: self.discrete_vals[i] = unique

        # 2. Stage 1: GBDT Training
        print("Step 1: Training GBDT Base Models...")
        base_preds = np.zeros_like(data)
        for col in tqdm(range(n_features)):
            valid_cols = [i for i in range(n_features) if i != col]
            model = lgb.LGBMRegressor(verbosity=-1, n_estimators=300, learning_rate=0.05)
            model.fit(data[:, valid_cols], data[:, col])
            self.gbdt_models[col] = model
            base_preds[:, col] = model.predict(data[:, valid_cols])
            
        # 3. Stage 2: Diffusion Training
        print("Step 2: Training Residual Diffusion...")
        residuals = data - base_preds
        res_norm = self.res_scaler.fit_transform(residuals)
        cond_norm = self.cond_scaler.fit_transform(base_preds)
        
        dataset = TensorDataset(torch.FloatTensor(res_norm), torch.FloatTensor(cond_norm))
        loader = DataLoader(dataset, batch_size=64, shuffle=True)
        
        self.diffusion = ResidualDiffusion(n_features).to(self.device)
        self.diff_manager = DiffusionManager(self.diffusion, self.device, timesteps=300)
        optimizer = optim.AdamW(self.diffusion.parameters(), lr=2e-4)
        
        for epoch in range(200):
            for res_b, cond_b in loader:
                self.diff_manager.train_step(res_b.to(self.device), cond_b.to(self.device), optimizer)
                
        # 4. Stage 3: Refiner Training
        print("Step 3: Training Autoencoder Refiner...")
        self.refiner = Refiner(n_features, self.device)
        self.refiner.fit(data, epochs=150)

    def impute(self, data, mask, n_samples=100):
        n_rows, n_features = data.shape
        
        # 1. Base Preds
        print("Generating Base...")
        X_clean = data.copy()
        for i in range(n_features):
            col = X_clean[:, i]
            valid = col[col != -1]
            X_clean[col == -1, i] = np.mean(valid) if len(valid)>0 else 0
            
        base_preds = np.zeros_like(data)
        for col in range(n_features):
            valid_cols = [i for i in range(n_features) if i != col]
            base_preds[:, col] = self.gbdt_models[col].predict(X_clean[:, valid_cols])
            
        # 2. Diffusion Residuals
        print("Generating Residuals...")
        cond_norm = self.cond_scaler.transform(base_preds)
        cond_tensor = torch.FloatTensor(cond_norm).to(self.device)
        
        batch_size = 100
        all_res = []
        for i in range(0, len(cond_tensor), batch_size):
            batch_cond = cond_tensor[i:i+batch_size]
            with torch.no_grad():
                res_gen = self.diff_manager.generate(batch_cond, n_samples=n_samples)
            res_np = res_gen.cpu().numpy()
            curr_N = res_np.shape[0]
            res_flat = res_np.reshape(-1, n_features)
            res_inv_flat = self.res_scaler.inverse_transform(res_flat)
            all_res.append(res_inv_flat.reshape(curr_N, n_samples, n_features))
            
        gen_residuals = np.concatenate(all_res, axis=0)
        
        # Combine
        final_samples = base_preds[:, None, :] + gen_residuals
        
        # 3. Refine (Autoencoder)
        final_samples = self.refiner.refine(final_samples)
        
        # 4. Snap & Mask
        print("Final Snapping...")
        for col in range(n_features):
            if col in self.discrete_vals:
                valids = self.discrete_vals[col]
                col_data = final_samples[:, :, col].flatten()
                diff = np.abs(col_data[:, None] - valids[None, :])
                snapped = valids[diff.argmin(axis=1)]
                final_samples[:, :, col] = snapped.reshape(n_rows, n_samples)
            
            observed_idx = np.where(mask[:, col] == 1)[0]
            final_samples[observed_idx, :, col] = data[observed_idx, col][:, None]
            
        return final_samples

# ==========================================
# 4. EXECUTION
# ==========================================
splits = load_dataset_splits('dataset')
train_original = splits['train']['original']
val_original = splits['val']['original']
X_full = np.vstack([train_original, val_original])
mask_full = np.ones_like(X_full)

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using device: {device}")

# A. Train
pipeline = TriStageImputer(device)
pipeline.fit(X_full, mask_full)

# B. Eval
print("\nINTERNAL EVALUATION")
test_imputed = splits['test']['imputed']
test_original = splits['test']['original']
test_mask = (test_imputed != -1).astype(int)

eval_samples = pipeline.impute(test_imputed, test_mask, n_samples=100)
test_mean = eval_samples.mean(axis=1)

mse_scores = []
for i in range(X_full.shape[1]):
    missing = np.where(test_mask[:, i] == 0)[0]
    if len(missing) > 0:
        mse_scores.append(mean_squared_error(test_original[missing, i], test_mean[missing, i]))
print(f"Eval MSE: {np.mean(mse_scores):.4f}")

plot_distribution_comparison(test_mean, test_original, test_mask, feature_names=[f'F{i+1}' for i in range(X_full.shape[1])])
plot_prediction_scatter(test_mean, test_original, test_mask, feature_names=[f'F{i+1}' for i in range(X_full.shape[1])])

# C. Submission
print("\nGenerating Submission...")
X_test2 = splits['test2']['imputed']
mask_test2 = (X_test2 != -1).astype(int)
test2_samples = pipeline.impute(X_test2, mask_test2, n_samples=100)
np.save('submission_tri_stage_v26.npy', test2_samples)
print("Saved submission_tri_stage_v26.npy")
```
