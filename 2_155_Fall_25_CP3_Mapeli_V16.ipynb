## Understanding the Data  
You are given thousands of anonymized and normalised numeric feature vectors representing real car designs.  

However, the team remembers that the features originally came from categories like:

- **Physical geometric parameters**  
  Length, ramp angles, bumper curvature, roof curvature, panel slopes, hood angle, etc.  
  *(But you won’t know which feature corresponds to which.)*

- **Aerodynamic coefficients**  
  Drag coefficient (Cd), lift/downforce (Cl), and other flow-derived metrics.

- **Cabin and packaging descriptors**  
  Approximate cabin volume, frontal area, interior shape metrics.

Your model must learn correlations between them to generate valid completions.

To simulate real engineering constraints, **some features are revealed** (the known physics/performance requirements) and others are **masked**.  
Your AI Copilot must generate **many plausible completions** for these masked (free) parameters.

1. **Train a generative model** (VAE, diffusion, CVAE, masked autoencoder, etc.) on the anonymized feature vectors.

3. Use your model to generate **multiple diverse, feasible completions** for the free parameters.  
4. Ensure that your generated designs:  
   - **Satisfy the known constraints**  
   - **Lie in the valid data manifold** (satisfy the conditional distribution of the free vs constrained parameters)  
   - **Are diverse** (many different feasible designs, not one solution)

### Data Preprocessing and Missing Value Handling

This is a critical section where we prepare our data for the VAE model. Key points:

**Missing Value Conventions:**
- In CSV files: `-1` indicates missing values
- In mask files: `True` = missing, `False` = observed
- For PyTorch models: We convert to `1` = observed, `0` = missing (standard convention)

**Why This Matters:**
Our VAE needs to distinguish between observed values (which provide constraints) and missing values (which need to be generated). The mask tells the model which values to trust and which to predict.

```python
# Correcting skeweed distributions
class FeatureTransformer:
    """
    Handles preprocessing for car design features, specifically applying the 
    Yeo-Johnson transformation to skewed features (28-34) to make them more Gaussian.
    
    The transformer is fitted only on non-missing data to ensure robustness.
    """
    def __init__(self, feature_indices_to_transform: List[int]):
        """
        Initializes the transformer.
        :param feature_indices_to_transform: A list of 0-indexed column indices 
                                             corresponding to the skewed features.
        """
        self.indices = feature_indices_to_transform
        # Use a list of transformers, one for each feature
        self.transformers: List[PowerTransformer] = [
            PowerTransformer(method='yeo-johnson', standardize=True) 
            for _ in feature_indices_to_transform
        ]
        print(f"Transformer initialized for {len(self.indices)} features.")

    def fit(self, X_train_NaN: pd.DataFrame):
        """
        Fits the PowerTransformer to the non-missing training data.
        
        :param X_train_NaN: The training data as a DataFrame, with missing values 
                            represented by NaN.
        """
        if not self.indices:
            print("No features configured for transformation.")
            return

        print(f"Fitting PowerTransformer on non-missing data for indices: {self.indices}...")
        
        # 1. Select the columns that need transformation
        cols_to_transform = X_train_NaN.iloc[:, self.indices]
        
        # 2. Iterate through each feature and fit its dedicated transformer
        for i, col_index in enumerate(self.indices):
            feature_data = cols_to_transform.iloc[:, i].dropna().values.reshape(-1, 1)
            
            if len(feature_data) > 0:
                self.transformers[i].fit(feature_data)
                
        print("✓ PowerTransformer fitted successfully.")

    def transform(self, X_NaN: Union[pd.DataFrame, np.ndarray]) -> np.ndarray:
        """
        Applies the fitted transformation to the input data.
        
        :param X_NaN: The data (train, validation, or test) with NaN for missing values.
        :return: A NumPy array with the specified features transformed.
        """
        if not self.indices:
            return X_NaN.values if isinstance(X_NaN, pd.DataFrame) else X_NaN
            
        X_out = X_NaN.values.copy() if isinstance(X_NaN, pd.DataFrame) else X_NaN.copy()
        
        print(f"Transforming data for indices: {self.indices}...")
        
        for i, col_index in enumerate(self.indices):
            # Extract the column data (including NaNs)
            col_data = X_out[:, col_index].reshape(-1, 1)
            
            # Identify where the NaNs are
            nan_mask = np.isnan(col_data).flatten()
            
            # Only transform non-NaN values
            if np.sum(~nan_mask) > 0:
                # Apply the transformation
                transformed_values = self.transformers[i].transform(col_data[~nan_mask])
                
                # Place transformed values back into the output array
                X_out[~nan_mask, col_index] = transformed_values.flatten()

        return X_out
    
```

```python
# Features 28-34 correspond to 0-indexed positions 27 through 33
SKEWED_INDICES = list(range(27, 34)) 

X_train_original_c = X_train_original.copy()

df_X_train_original_c = pd.DataFrame(X_train_original_c)

# 2. Initialize and fit the transformer
preprocessor = FeatureTransformer(SKEWED_INDICES)
preprocessor.fit(df_X_train_original_c)
    
# 3. Apply the transformation
transformed_data_np = preprocessor.transform(df_X_train_original_c)
transformed_df = pd.DataFrame(transformed_data_np)
    
# 4. Check the results (Should be close to zero mean, close to zero median)
print("\n--- Transformed Skewness Check (Feature 28, index 27) ---")
print(f"Transformed Mean: {transformed_df.iloc[:, 27].mean():.3f}")
print(f"Transformed Median: {transformed_df.iloc[:, 27].median():.3f}")
    
# Note: NaN values remain NaN after transformation
print(f"NaN count after transform: {np.isnan(transformed_data_np[:, 27]).sum()}")
```

```python
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset
from sklearn.preprocessing import StandardScaler, QuantileTransformer
from sklearn.metrics import mean_squared_error
from sklearn.neighbors import NearestNeighbors
from sklearn.utils import resample
import lightgbm as lgb
from tqdm import tqdm
import matplotlib.pyplot as plt
import math
import warnings
import os

# --- UTILS ---
def load_dataset_splits(data_dir='dataset'):
    splits = {}
    split_types = ['train', 'val', 'test', 'test2']
    print(f"Loading dataset splits from: {data_dir}")
    for split_type in split_types:
        print(f"Loading {split_type} split...")
        splits[split_type] = {}
        if split_type != 'test2':
            original_path = os.path.join(data_dir, f'{split_type}_original.csv')
            if os.path.exists(original_path):
                splits[split_type]['original'] = pd.read_csv(original_path).values
        imputed_path = os.path.join(data_dir, f'{split_type}_imputed.csv')
        if os.path.exists(imputed_path):
            splits[split_type]['imputed'] = pd.read_csv(imputed_path).values
        mask_path = os.path.join(data_dir, f'{split_type}_missing_mask.csv')
        if os.path.exists(mask_path):
            splits[split_type]['missing_mask'] = pd.read_csv(mask_path).values.astype(bool)
    return splits

def plot_full_evaluation(predictions, ground_truth, imputed_input, feature_names):
    """Restored Full Pagination Logic"""
    # Derive masks
    real_mask = (imputed_input != -1).astype(bool)
    missing_mask = ~real_mask 
    n_features = predictions.shape[1]
    chunk_size = 16
    
    for start_idx in range(0, n_features, chunk_size):
        end_idx = min(start_idx + chunk_size, n_features)
        current_features = range(start_idx, end_idx)
        n_current = len(current_features)
        
        n_cols = 4
        n_rows = (n_current + n_cols - 1) // n_cols
        
        # Distribution Plots
        fig, axes = plt.subplots(n_rows, n_cols * 2, figsize=(20, 4 * n_rows))
        fig.suptitle(f"Distributions (F{start_idx+1}-F{end_idx})", fontsize=16, y=1.02)
        axes = axes.flatten()
        
        for i, feat_idx in enumerate(current_features):
            idx = missing_mask[:, feat_idx]
            ax_gt = axes[2*i]
            ax_pred = axes[2*i+1]
            
            if idx.sum() == 0: continue
            gt = ground_truth[idx, feat_idx]
            pred = predictions[idx, feat_idx]
            
            all_d = np.concatenate([gt, pred])
            vmin, vmax = all_d.min(), all_d.max()
            
            ax_gt.hist(gt, bins=20, density=True, color='skyblue', alpha=0.8)
            ax_gt.set_title(f"{feature_names[feat_idx]} GT")
            ax_gt.set_xlim(vmin, vmax)
            
            ax_pred.hist(pred, bins=20, density=True, color='salmon', alpha=0.8)
            ax_pred.set_title(f"{feature_names[feat_idx]} Model")
            ax_pred.set_xlim(vmin, vmax)
        plt.tight_layout()
        plt.show()
        
        # Scatter Plots
        fig, axes = plt.subplots(n_rows, n_cols, figsize=(20, 5 * n_rows))
        fig.suptitle(f"Accuracy (F{start_idx+1}-F{end_idx})", fontsize=16, y=1.02)
        axes = axes.flatten()
        for i, feat_idx in enumerate(current_features):
            idx = missing_mask[:, feat_idx]
            ax = axes[i]
            if idx.sum() < 2: continue
            gt = ground_truth[idx, feat_idx]
            pred = predictions[idx, feat_idx]
            ax.scatter(gt, pred, alpha=0.3, s=10)
            vmin, vmax = min(gt.min(), pred.min()), max(gt.max(), pred.max())
            ax.plot([vmin, vmax], [vmin, vmax], 'r--')
            mse = mean_squared_error(gt, pred)
            ax.set_title(f"{feature_names[feat_idx]}\nMSE: {mse:.4f}")
        plt.tight_layout()
        plt.show()

warnings.filterwarnings('ignore')

# ==========================================
# 1. MANIFOLD AUTOENCODER (The Refiner)
# ==========================================
class Autoencoder(nn.Module):
    def __init__(self, input_dim, hidden_dims=[128, 64, 32]):
        super().__init__()
        layers = []
        in_d = input_dim
        for h in hidden_dims:
            layers.extend([nn.Linear(in_d, h), nn.BatchNorm1d(h), nn.LeakyReLU(), nn.Dropout(0.05)])
            in_d = h
        self.encoder = nn.Sequential(*layers)
        layers = []
        for h in reversed(hidden_dims):
            layers.extend([nn.Linear(in_d, h), nn.BatchNorm1d(h), nn.LeakyReLU(), nn.Dropout(0.05)])
            in_d = h
        self.decoder = nn.Sequential(*layers)
        self.final = nn.Linear(in_d, input_dim)
        
    def forward(self, x):
        z = self.encoder(x)
        return self.final(self.decoder(z))

class ManifoldRefiner:
    def __init__(self, input_dim, device):
        self.device = device
        self.model = Autoencoder(input_dim).to(device)
        self.scaler = StandardScaler()
        
    def fit(self, data, epochs=150):
        print("Training Manifold Refiner...")
        data_norm = self.scaler.fit_transform(data)
        loader = DataLoader(TensorDataset(torch.FloatTensor(data_norm)), batch_size=64, shuffle=True)
        optimizer = optim.Adam(self.model.parameters(), lr=1e-3)
        for epoch in range(epochs):
            self.model.train()
            for batch in loader:
                x = batch[0].to(self.device)
                noise = torch.randn_like(x) * 0.1 
                recon = self.model(x + noise)
                loss = nn.MSELoss()(recon, x)
                optimizer.zero_grad()
                loss.backward()
                optimizer.step()

    def refine(self, samples):
        print("Refining...")
        n_samples, n_variations, n_features = samples.shape
        flat = samples.reshape(-1, n_features)
        refined_flat = np.zeros_like(flat)
        self.model.eval()
        batch_size = 1000
        for i in range(0, len(flat), batch_size):
            batch = flat[i:i+batch_size]
            batch_norm = self.scaler.transform(batch)
            batch_t = torch.FloatTensor(batch_norm).to(self.device)
            with torch.no_grad():
                recon = self.model(batch_t)
                # TWEAK: 0.6 GBDT + 0.4 AE
                refined = 0.6 * batch_t + 0.4 * recon
            refined_flat[i:i+batch_size] = self.scaler.inverse_transform(refined.cpu().numpy())
        return refined_flat.reshape(n_samples, n_variations, n_features)

# ==========================================
# 2. DIFFUSION MODEL
# ==========================================
class SinusoidalPositionEmbeddings(nn.Module):
    def __init__(self, dim):
        super().__init__()
        self.dim = dim
    def forward(self, time):
        device = time.device
        half_dim = self.dim // 2
        embeddings = math.log(10000) / (half_dim - 1)
        embeddings = torch.exp(torch.arange(half_dim, device=device) * -embeddings)
        embeddings = time[:, None] * embeddings[None, :]
        return torch.cat((embeddings.sin(), embeddings.cos()), dim=-1)

class ResidualDiffusion(nn.Module):
    def __init__(self, input_dim):
        super().__init__()
        self.time_mlp = nn.Sequential(
            SinusoidalPositionEmbeddings(64),
            nn.Linear(64, 64), nn.SiLU(), nn.Linear(64, 64)
        )
        in_dim = (input_dim * 2) + 64
        self.layers = nn.ModuleList([
            nn.Sequential(nn.Linear(in_dim, 512), nn.BatchNorm1d(512), nn.SiLU(), nn.Dropout(0.1)),
            nn.Sequential(nn.Linear(512, 512), nn.BatchNorm1d(512), nn.SiLU(), nn.Dropout(0.1)),
            nn.Sequential(nn.Linear(512, 512), nn.BatchNorm1d(512), nn.SiLU(), nn.Dropout(0.1))
        ])
        self.final = nn.Linear(512, input_dim)
    def forward(self, x, t, cond):
        t_emb = self.time_mlp(t)
        x = torch.cat([x, cond, t_emb], dim=1)
        for l in self.layers: x = l(x)
        return self.final(x)

class DiffusionManager:
    def __init__(self, model, device, timesteps=500): # 500 Steps
        self.model = model
        self.device = device
        self.timesteps = timesteps
        self.betas = torch.linspace(0.0001, 0.02, timesteps).to(device)
        self.alphas = 1. - self.betas
        self.alphas_cumprod = torch.cumprod(self.alphas, dim=0)
        self.sqrt_alphas_cumprod = torch.sqrt(self.alphas_cumprod)
        self.sqrt_one_minus_alphas_cumprod = torch.sqrt(1. - self.alphas_cumprod)
    def add_noise(self, x_0, t):
        noise = torch.randn_like(x_0)
        sqrt_alpha = self.sqrt_alphas_cumprod[t][:, None]
        sqrt_one_minus = self.sqrt_one_minus_alphas_cumprod[t][:, None]
        return sqrt_alpha * x_0 + sqrt_one_minus * noise, noise
    def train_step(self, res, cond, optimizer):
        self.model.train()
        optimizer.zero_grad()
        t = torch.randint(0, self.timesteps, (res.shape[0],), device=self.device).long()
        x_t, noise = self.add_noise(res, t)
        pred_noise = self.model(x_t, t, cond)
        loss = nn.MSELoss()(pred_noise, noise)
        loss.backward()
        optimizer.step()
        return loss.item()
    @torch.no_grad()
    def generate(self, cond, n_samples=1):
        self.model.eval()
        n, f = cond.shape
        cond_rep = cond.repeat_interleave(n_samples, dim=0)
        x = torch.randn_like(cond_rep)
        for i in reversed(range(self.timesteps)):
            t = torch.full((x.shape[0],), i, device=self.device, dtype=torch.long)
            pred_noise = self.model(x, t, cond_rep)
            beta = self.betas[i]
            alpha = self.alphas[i]
            alpha_bar = self.alphas_cumprod[i]
            mean = (1 / torch.sqrt(alpha)) * (x - (beta / torch.sqrt(1 - alpha_bar)) * pred_noise)
            if i > 0:
                noise = torch.randn_like(x)
                sigma = torch.sqrt(beta)
                x = mean + sigma * noise
            else: x = mean
        return x.reshape(n, n_samples, f)

# ==========================================
# 3. HYBRID IMPUTER (With Bagging Logic)
# ==========================================
class ResidualHybridImputer:
    def __init__(self, device):
        self.device = device
        self.gbdt_models = {} # Will store lists of models for bagging
        self.diffusion = None
        self.diff_manager = None
        self.res_scaler = QuantileTransformer(output_distribution='normal', n_quantiles=2000)
        self.cond_scaler = StandardScaler()
        self.discrete_vals = {}
        self.n_bags = 200 # Increased Bags
    
    def fit(self, data, mask):
        n, f = data.shape
        for i in range(f):
            u = np.unique(data[:, i])
            if len(u) < 100: self.discrete_vals[i] = u
            
        print(f"Training GBDT Base ({self.n_bags} Bags)...")
        # Initialize lists for bagging
        for col in range(f): self.gbdt_models[col] = []
            
        # Train Bags
        for b in tqdm(range(self.n_bags)):
            # Bootstrap
            indices = np.arange(n)
            boot_idx = resample(indices, replace=True, random_state=42+b)
            X_boot = data[boot_idx]
            
            # Simple mean prediction to collect residuals for diffusion later
            # We only need one "best guess" for diffusion training residuals, 
            # but we need 200 models for inference diversity.
            
            for col in range(f):
                valid = [i for i in range(f) if i!=col]
                model = lgb.LGBMRegressor(verbosity=-1, n_estimators=100, learning_rate=0.05, random_state=b)
                model.fit(X_boot[:, valid], X_boot[:, col])
                self.gbdt_models[col].append(model)
        
        # For diffusion training, we need ONE set of residuals.
        # We'll use the average prediction of the first 5 bags to keep it stable.
        print("Training Diffusion...")
        base_preds = np.zeros_like(data)
        for col in range(f):
            valid = [i for i in range(f) if i!=col]
            # Avg first 5 models
            preds_accum = np.zeros(n)
            for m in self.gbdt_models[col][:5]:
                preds_accum += m.predict(data[:, valid])
            base_preds[:, col] = preds_accum / 5
            
        residuals = data - base_preds
        res_n = self.res_scaler.fit_transform(residuals)
        cond_n = self.cond_scaler.fit_transform(base_preds)
        loader = DataLoader(TensorDataset(torch.FloatTensor(res_n), torch.FloatTensor(cond_n)), batch_size=64, shuffle=True)
        
        self.diffusion = ResidualDiffusion(f).to(self.device)
        self.diff_manager = DiffusionManager(self.diffusion, self.device, timesteps=500)
        opt = optim.AdamW(self.diffusion.parameters(), lr=2e-4)
        
        for epoch in range(200):
            for r, c in loader:
                self.diff_manager.train_step(r.to(self.device), c.to(self.device), opt)

    def impute(self, data, mask, n_samples=100):
        n, f = data.shape
        print("Generating Base (Bagging)...")
        X_c = data.copy()
        for i in range(f):
            c = X_c[:, i]
            v = c[c != -1]
            X_c[c == -1, i] = np.mean(v) if len(v)>0 else 0
            
        # Bagging: Cycle through models to generate diversity in Base Prediction
        base_preds_all = np.zeros((n, n_samples, f))
        
        for s in tqdm(range(n_samples)):
            bag_idx = s % self.n_bags
            for col in range(f):
                valid = [i for i in range(f) if i!=col]
                model = self.gbdt_models[col][bag_idx]
                base_preds_all[:, s, col] = model.predict(X_c[:, valid])
                
        # Flatten for diffusion
        base_flat = base_preds_all.reshape(-1, f)
            
        print("Generating Residuals...")
        cond_n = self.cond_scaler.transform(base_flat)
        cond_t = torch.FloatTensor(cond_n).to(self.device)
        
        res_list = []
        bs = 500
        for i in range(0, len(cond_t), bs):
            batch = cond_t[i:i+bs]
            with torch.no_grad():
                # Generate 1 residual per base prediction
                gen = self.diff_manager.generate(batch, n_samples=1)
            gen_np = gen.cpu().numpy()
            # Shape (Batch, 1, F) -> (Batch, F)
            res_inv = self.res_scaler.inverse_transform(gen_np.reshape(-1, f))
            res_list.append(res_inv)
            
        gen_residuals = np.concatenate(res_list, axis=0)
        final_flat = base_flat + gen_residuals
        final = final_flat.reshape(n, n_samples, f)
        
        print("Restoring Observed...")
        for col in range(f):
            obs = np.where(mask[:, col] == 1)[0]
            final[obs, :, col] = data[obs, col][:, None]
        return final

# ==========================================
# 4. GRAPH SMOOTHER
# ==========================================
class GraphSmoother:
    def __init__(self, n_neighbors=50, alpha=0.3):
        self.knn = NearestNeighbors(n_neighbors=n_neighbors, n_jobs=-1)
        self.alpha = alpha
        
    def smooth(self, samples, mask, orig):
        print("Smoothing...")
        n, v, f = samples.shape
        smoothed = samples.copy()
        means = samples.mean(axis=1)
        self.knn.fit(means)
        dists, idxs = self.knn.kneighbors(means)
        weights = 1.0 / (dists + 1e-6)
        weights /= weights.sum(axis=1, keepdims=True)
        
        for s in tqdm(range(v)):
            world = samples[:, s, :]
            neighbors = world[idxs]
            avg = (weights[:, :, None] * neighbors).sum(axis=1)
            new_world = (1 - self.alpha) * world + self.alpha * avg
            for c in range(f):
                obs = np.where(mask[:, c] == 1)[0]
                new_world[obs, c] = orig[obs, c]
            smoothed[:, s, :] = new_world
        return smoothed

# ==========================================
# 5. EXECUTION
# ==========================================
if __name__ == "__main__":
    splits = load_dataset_splits('dataset')
    X_train = splits['train']['original']
    X_val = splits['val']['original']
    X_full = np.vstack([X_train, X_val])
    mask_full = np.ones_like(X_full)
    
    device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    print(f"Using {device}")
    
    # 1. Train
    imputer = ResidualHybridImputer(device)
    imputer.fit(X_full, mask_full)
    refiner = ManifoldRefiner(X_full.shape[1], device)
    refiner.fit(X_full, epochs=200)
    
    # 2. Eval
    print("\nEVALUATION")
    test_imputed = splits['test']['imputed']
    test_orig = splits['test']['original']
    test_mask = (test_imputed != -1).astype(int)
    
    samples = imputer.impute(test_imputed, test_mask)
    samples = GraphSmoother().smooth(samples, test_mask, test_imputed)
    samples = refiner.refine(samples)
    
    # Snap
    for c in range(X_full.shape[1]):
        if c in imputer.discrete_vals:
            valid = imputer.discrete_vals[c]
            for s in range(100):
                diff = np.abs(samples[:, s, c][:, None] - valid[None, :])
                samples[:, s, c] = valid[diff.argmin(axis=1)]
                
    mse = []
    mean_pred = samples.mean(axis=1)
    for c in range(X_full.shape[1]):
        m = np.where(test_mask[:, c] == 0)[0]
        if len(m) > 0: mse.append(mean_squared_error(test_orig[m, c], mean_pred[m, c]))
    print(f"MSE: {np.mean(mse):.4f}")
    
    plot_full_evaluation(mean_pred, test_orig, test_imputed, [f'F{i+1}' for i in range(X_full.shape[1])])
    
    # 3. Submission
    print("\nSUBMISSION")
    X_sub = splits['test2']['imputed']
    mask_sub = (X_sub != -1).astype(int)
    
    sub_samples = imputer.impute(X_sub, mask_sub)
    sub_samples = GraphSmoother().smooth(sub_samples, mask_sub, X_sub)
    sub_samples = refiner.refine(sub_samples)
    
    for c in range(X_full.shape[1]):
        if c in imputer.discrete_vals:
            valid = imputer.discrete_vals[c]
            for s in range(100):
                diff = np.abs(sub_samples[:, s, c][:, None] - valid[None, :])
                sub_samples[:, s, c] = valid[diff.argmin(axis=1)]
                
    np.save('submission_final_v30.npy', sub_samples)
    print("Done.")
```
